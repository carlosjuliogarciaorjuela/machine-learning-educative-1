A. Introduction to linear regression
One of the main objectives in both machine learning and data science is finding an equation or distribution that best fits a given dataset. This is known as data modeling, where we create a model that uses the dataset’s features as independent variables to predict output values for some dependent variable (with minimal error).

However, it is incredibly difficult to find an optimal model for most datasets, given the amount of noise (i.e. random errors/fluctuations) in real world data.

Since finding an optimal model for a dataset is difficult, we instead try to find a good approximating distribution. In many cases, a linear model (a linear combination of the dataset’s features) can approximate the data well. The term linear regression refers to using a linear model to represent the relationship between a set of independent variables and a dependent variable.

B. Basic linear regression
The simplest form of linear regression is called least squares regression. This strategy produces a regression model, which is a linear combination of the independent variables, that minimizes the sum of squared residuals between the model’s predictions and actual values for the dependent variable.

Implementing basic linear regression with scikit-learn
In scikit-learn, the least squares regression model is implemented with the LinearRegression object, which is a part of the linear_model module in sklearn. The object contains a fit function, which takes in an input dataset of features (independent variables) and an array of labels (dependent variables) for each data observation (rows of the dataset). The code below demonstrates how to fit a LinearRegression model to a dataset of 5 different pizzas (pizza_data) and corresponding pizza prices. The first column of pizza_data represents the number of calories and the second column represents net weight (in grams).

```python
# Importamos librerias
import numpy as np
from sklearn import linear_model

# Diámetro de la pizza en pulgadas (ejemplo de característica / X)
# Debe ser una matriz de 2D: cada fila es una muestra
pizza_data = np.array([[6], [8], [10], [14], [18]])

# Precio de la pizza en dólares (ejemplo de etiqueta / y)
# Puede ser un arreglo de 1D o 2D
pizza_prices = np.array([7, 9, 13, 17.5, 18])

# Verificamos los datos
print('{}\n'.format(repr(pizza_data)))
print('{}\n'.format(repr(pizza_prices)))

# Ahora ejecutas tu modelo de Scikit-Learn
reg = linear_model.LinearRegression()
reg.fit(pizza_data, pizza_prices)

print("Entrenamiento completado con éxito.")
```

After calling the fit function, the model is ready to use. The predict function allows us to make predictions on new data. We can also get the specific coefficients and intercept for the linear combination using the coef_ and intercept_ properties, respectively.
```python
# 1. Conjunto de datos: Diámetros (pulgadas) y Precios (dólares)
pizza_data = np.array([[6], [8], [10], [14], [18]])
pizza_prices = np.array([7, 9, 13, 17.5, 18])

# 2. Creación y entrenamiento del modelo
reg = linear_model.LinearRegression()
reg.fit(pizza_data, pizza_prices)

# 3. Nuevos datos para predecir (pizzas de 12 y 16 pulgadas)
new_pizzas = np.array([[12], [16]])
price_predicts = reg.predict(new_pizzas)

# 4. Resultados de las predicciones
print('Predicciones para 12" y 16":')
print(repr(price_predicts))
print('-' * 40)

# 5. Parámetros aprendidos por la recta (y = m*x + b)
print('Pendiente (coef_): {}'.format(reg.coef_))
print('Intersección (intercept_): {:.2f}'.format(reg.intercept_))
print('-' * 40)

# 6. Evaluación de calidad del modelo (R²)
r2 = reg.score(pizza_data, pizza_prices)
print('Score R2: {:.4f}'.format(r2))
```

C. When to use linear regression
Linear regression is a powerful yet simple tool for predicting continuous outcomes. It works best when the relationship between the independent variables and the dependent variable is roughly linear, and the dataset is relatively clean.

Use linear regression in the following scenarios:

Linear relationships: When features have a proportional effect on the target.

Predictive modeling: Estimating numerical outcomes such as sales, prices, or temperatures.

Baseline modeling: As a starting point before trying more complex models.

Trend analysis: Understanding how variables impact the target over time.

Low noise datasets: Works best when outliers or extreme values are minimal.

D. Advantages of linear regression
Linear regression is widely used due to its simplicity, interpretability, and efficiency. It also provides insight into the impact of each independent variable on the target, making it useful for both prediction and analysis.
